```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df=pd.read_csv("/content/employee_salary_ml_practice.csv")
df.head()
df.info()
df=df.drop(columns=["Name"])
print(df)
```

```python
X= df.drop(columns=["Salary"])
y = df["Salary"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=0)
print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
numeric_features=X_train.select_dtypes(include=["int64","float64"]).columns.tolist()
catagorical_features=X_train.select_dtypes(include=["object"]).columns.tolist()
print("\nNumeric features:", numeric_features)
print("Catagorical features:", catagorical_features)
```

```python
numeric_transformer=Pipeline([("Imputer",SimpleImputer(strategy="median")),("scaler",StandardScaler())])

catagorical_transformer=Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("encoder",OneHotEncoder(handle_unknown="ignore"))])

preprocessor=ColumnTransformer([("num",numeric_transformer,numeric_features),("cat",catagorical_transformer,catagorical_features)])
```

```python
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        random_state=0
    ))
])


# 11. Train model
rf_model.fit(X_train, y_train)


# 12. Prediction
y_pred_rf = rf_model.predict(X_test)


# 13. Evaluation
mae = mean_absolute_error(y_test, y_pred_rf)
mse = mean_squared_error(y_test, y_pred_rf)
rmse = mean_squared_error(y_test, y_pred_rf) ** 0.5
r2 = r2_score(y_test, y_pred_rf)


print("\nModel Evaluation")
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R² Score:", round(r2, 3))
```
